```python
import pandas as pd
from pathlib import Path

folder = Path(r"C:\Users\jclin\Downloads\Maven+Fuzzy+Factory")

csv_files = list(folder.glob("*.csv"))

for file in csv_files:
    print(file.name)
```

```python

orders = pd.read_csv(folder / "orders.csv")
order_items = pd.read_csv(folder / "order_items.csv")
order_item_refunds = pd.read_csv(folder / "order_item_refunds.csv")
products = pd.read_csv(folder / "products.csv")
website_pageviews = pd.read_csv(folder / "website_pageviews.csv")
website_sessions = pd.read_csv(folder / "website_sessions.csv")
```

```python
datasets = {
    "orders": orders,
    "order_items": order_items,
    "order_item_refunds": order_item_refunds,
    "products": products,
    "website_pageviews": website_pageviews,
    "website_sessions": website_sessions
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows x {df.shape[1]} columns")
```

```python
primary_keys = {
    "orders": "order_id",
    "order_items": "order_item_id",
    "order_item_refunds": "order_item_refund_id",
    "products": "product_id",
    "website_pageviews": "website_pageview_id",
    "website_sessions": "website_session_id"
}

for name, df in datasets.items():
    pk = primary_keys[name]

    print(f"\n{'='*70}")
    print(name.upper())
    print(f"{'='*70}")

    print(f"Total rows:              {len(df):,}")
    print(f"Fully duplicated rows:   {df.duplicated().sum():,}")
    print(f"Unique {pk}: {df[pk].nunique():,}")
    print(f"Duplicate {pk}: {df[pk].duplicated().sum():,}")
```

```python

checks = {
    "orders.website_session_id -> website_sessions.website_session_id":
        ~orders["website_session_id"].isin(website_sessions["website_session_id"]),

    "order_items.order_id -> orders.order_id":
        ~order_items["order_id"].isin(orders["order_id"]),

    "order_items.product_id -> products.product_id":
        ~order_items["product_id"].isin(products["product_id"]),

    "order_item_refunds.order_item_id -> order_items.order_item_id":
        ~order_item_refunds["order_item_id"].isin(order_items["order_item_id"]),

    "order_item_refunds.order_id -> orders.order_id":
        ~order_item_refunds["order_id"].isin(orders["order_id"]),

    "website_pageviews.website_session_id -> website_sessions.website_session_id":
        ~website_pageviews["website_session_id"].isin(website_sessions["website_session_id"])
}

for relationship, invalid_mask in checks.items():
    print(f"{relationship}: {invalid_mask.sum():,} unmatched rows")
```

```python
# Date ranges
print("DATE RANGES")
print("-" * 60)

for name, df in datasets.items():
    if "created_at" in df.columns:
        dates = pd.to_datetime(df["created_at"])
        print(
            f"{name}: {dates.min()} -> {dates.max()}"
        )

# Categorical / flag values
print("\nCATEGORICAL VALUES")
print("-" * 60)

columns_to_check = {
    "website_sessions": [
        "is_repeat_session",
        "utm_source",
        "utm_campaign",
        "utm_content",
        "device_type"
    ],
    "order_items": [
        "is_primary_item"
    ],
    "website_pageviews": [
        "pageview_url"
    ]
}

for table_name, columns in columns_to_check.items():
    df = datasets[table_name]

    print(f"\n{table_name.upper()}")

    for col in columns:
        print(f"\n{col}:")
        print(df[col].value_counts(dropna=False).head(20))

# Numeric ranges
print("\nNUMERIC RANGES")
print("-" * 60)

numeric_checks = [
    ("orders", orders, ["items_purchased", "price_usd", "cogs_usd"]),
    ("order_items", order_items, ["price_usd", "cogs_usd"]),
    ("order_item_refunds", order_item_refunds, ["refund_amount_usd"])
]

for name, df, cols in numeric_checks:
    print(f"\n{name.upper()}")

    for col in cols:
        print(
            f"{col}: "
            f"min={df[col].min()}, "
            f"max={df[col].max()}, "
            f"mean={df[col].mean():.2f}"
        )
```
